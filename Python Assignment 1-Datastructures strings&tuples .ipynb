#2.STRING SLICING & INDEXING

```python
# Print the first character of the string
print("First character:",Final[0])
```

```python
# Print Last Character of the String
print("Last character:",Final[-1])
```

```python
# Print the first five character of the string
print("First five character of the string: ", Final [:5])
```

```python
# Print the last 11 characters of the string
print("Last eleven characters of the string: ",Final[:-11])
```

3.STRING METHODS

```python
#Convert the sentence to uppercase.
strM="Python beginner tutorial"
print("Uppercase: ", strM.upper() )
```

```python
#Convert the sentence into Lower case
print("Lowercase: ",strM.lower())
```

```python
#Use capitalize and reverse the sentence to original input form
print("capitalize:",strM.upper().capitalize())
```

```python
#count the total number occurence of character 't' in the string
print("count of 't':",strM.count('t'))
```

```python
#replace all occurence of "Python" with "Machine Learning"
print("Replace:",strM.replace("Python","Machine Learning"))
```
